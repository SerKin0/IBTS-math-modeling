Функция для отбора данных из таблицы данных.

Нам нужно взять информацию из столбцов, где нет значения «999.9». Поэтому мы сделаем так, чтобы строки с этим значением не учитывались.

```python
def clear_cols(
    array: np.ndarray, col: int | list[int] = None, error: float | int | list = 999.9
) -> list[float] | np.ndarray:
    if col is None:
        col = list(range(array.shape[1]))
    elif isinstance(col, int):
        col = [col]

    if isinstance(error, (float, int)):
        error = [error]

    mask = np.ones(len(array), dtype=bool)
    for c in col:
        for err in error:
            mask &= array[:, c] != err

    result = array[mask][:, col].T

    if len(col) == 1:
        return result[0]
    else:
        return result
```

### **Интерполяционный полином Ньютона (первая или вперед)**
$$\Delta^k y_i = y_{i+k} - k \cdot y_{i+k-1} + \frac{k(k-1)}{2!} y_{i+k-2} - \dots + (-1)^k y_i = \sum_{j=0}^k (-1)^{k-j} \begin{pmatrix}k \\j \end{pmatrix} y_{j}$$
$$x_{i+1} - x_i = h = \text{const} \qquad (i = 0\cdots n)$$
$$N_n(x) = y_0 + \frac{\Delta y_0}{h}(x - x_0) + \frac{\Delta^2 y_0}{2 h^2} (x - x_0)(x - x_1) + \dots + \frac{\Delta^n y_0}{n! h^n}(x - x_0) (x - x_1) \dots (x - x_{n-1})$$

```python
def delta_y_n(data: list[float] | np.ndarray, k: int, n: int = 0) -> float:
    if len(data) <= k:
        raise ValueError(
            f"Переменная k должна быть меньше длины списка: {k} >= {len(data)}"
        )

    if len(data) - k - 1 < n:
        raise ValueError(
            f"На строке {k=} не может быть больше {len(data) - k} элементов"
        )

    if k == 0:
        return data[0]

    result = 0
    b = 1.0
    for i in range(n, k + n + 1):
        result += (-1) ** (k - i + n) * b * data[i]
        b *= (k - i + n) / (i - n + 1)

    return result


def newton_interpolation_first(
    x: float, x_data: list[float] | np.ndarray, y_data: list[float] | np.ndarray
) -> float:
    h = x_data[1] - x_data[0]
    if not all(
        abs(x_data[i + 1] - x_data[i] - h) < 1e-10 for i in range(len(x_data) - 1)
    ):
        raise ValueError("Шаг по оси h различается")

    N = y_data[0]
    prod = 1.0
    fact = 1.0

    for i in range(1, len(y_data)):
        prod *= x - x_data[i - 1]
        fact *= i
        N += delta_y_n(y_data, i) * prod / (fact * h**i)

    return N
```

Получаем очищенные данные из необходимых столбцов. Пусть это будет *год* и *январь*.

```python
cols = (0, 1)

x_data, y_data = clear_cols(values, cols)

start = 0
end = start + 6

y_data, x_data = y_data[start:end], x_data[start:end]

print(f"{x_data=}")
print(f"{y_data=}")
```

Построение графика

```python
x_smooth = np.linspace(min(x_data), max(x_data), 100)
y_smooth = [newton_interpolation_first(x, x_data, y_data) for x in x_smooth]

plt.figure(figsize=(10, 6))
ax = plt.gca()
# ax.xaxis.set_minor_locator(MultipleLocator(0.1))
plt.grid(True, which="minor", linestyle=":", linewidth=0.5, color="black")

plt.plot(x_smooth, y_smooth, "--", label="Интерполяция Лагранжа")
plt.plot(x_data, y_data, "o", label="Исходные данные", markersize=8)
plt.xticks(np.arange(x_data[0], x_data[-1], 1), [i / 10 for i in range(x_data.shape[0]-1)])

plt.xlabel(headers[cols[0]])
plt.ylabel(headers[cols[1]])
plt.title("Интерполяция Лагранжа")
plt.legend()
plt.grid(True)
plt.show()
print(y_data)
```
