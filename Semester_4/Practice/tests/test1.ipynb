```python
#!py -3.8
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Создаем сетку координат x и y (уменьшаем количество точек для tikzplotlib)
x = np.linspace(-5, 5, 30)  # 30 точек вместо 100
y = np.linspace(-5, 5, 30)
X, Y = np.meshgrid(x, y)

# Функция z = x² + xy - x + y² - 5y + 4
Z = X**2 + X*Y - X + Y**2 - 5*Y + 4

# Создаем 3D график
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Отображаем поверхность
surf = ax.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)

# Подписи осей
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('График')

# Добавляем цветовую шкалу
fig.colorbar(surf)

# Сохраняем в tikz
import tikzplotlib
tikzplotlib.save("test.tex")
```
