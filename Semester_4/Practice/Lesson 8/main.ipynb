Вычислить интеграл по формуле трапеций с точностью $\epsilon = 10^{-4}$

$$
\boxed{\int_0^9 \frac{\ln (1+x)}{x\sqrt{x}}\, dx}
$$

Замена:
$$
t = \sqrt{x}, \quad x = t^2, \quad dx = 2t\, dt, \quad \text{пределы: } x=0 \to t=0, \ x=9 \to t=3
$$

Получаем:
$$
I = \int_0^3 \frac{\ln (1 + t^2)}{t^3} \cdot 2t\, dt = 2 \int_0^3 \frac{\ln (1 + t^2)}{t^2}\, dt
$$

Интегрирование по частям:
$$
u = \ln(1+t^2), \quad du = \frac{2t}{1+t^2}\, dt, \qquad dv = \frac{dt}{t^2}, \quad v = -\frac{1}{t}
$$

Тогда:
$$
I = 2 \left[ -\frac{\ln(1+t^2)}{t} \Bigg|_0^3 + \int_0^3 \frac{2}{1+t^2}\, dt \right]
$$

Граничные члены:
$$
-\frac{\ln(1+t^2)}{t} \Bigg|_0^3 = -\frac{\ln 10}{3} - \lim_{t \to 0^+} \left( -\frac{\ln(1+t^2)}{t} \right) = -\frac{\ln 10}{3}
$$
(так как $\lim_{t \to 0^+} \frac{\ln(1+t^2)}{t} = 0$)

Оставшийся интеграл:
$$
2 \cdot \int_0^3 \frac{2}{1+t^2}\, dt = 4 \int_0^3 \frac{dt}{1+t^2} = 4 \arctan t \Big|_0^3 = 4 \arctan 3
$$

Итоговый результат:
$$
I = 2 \left( -\frac{\ln 10}{3} + 2 \arctan 3 \right) = \boxed{-\frac{2\ln 10}{3} + 4 \arctan 3}
$$

Численное значение:
$$
\ln 10 \approx 2{,}302585, \quad -\frac{2 \cdot 2{,}302585}{3} \approx -1{,}535057
$$
$$
\arctan 3 \approx 1{,}249045, \quad 4 \cdot 1{,}249045 \approx 4{,}996180
$$
$$
I \approx -1{,}535057 + 4{,}996180 = \boxed{3{,}461123}
$$

Произведем замену переменных
$$
t = \sqrt{x}, \quad x = t^2, \quad dx = 2t\, dt, \quad \text{пределы: } x=0 \to t=0, \ x=9 \to t=3
$$
$$
I = \int_0^3 \frac{\ln(1+t^2)}{t^3} \cdot 2t\, dt = 2 \int_0^3 \frac{\ln(1+t^2)}{t^2}\, dt
$$

$$
u = \ln(1+t^2), \quad dv = \frac{dt}{t^2} \;\Rightarrow\; du = \frac{2t}{1+t^2}dt, \quad v = -\frac{1}{t}
$$

$$
I = 2\left[ -\frac{\ln(1+t^2)}{t}\Bigg|_0^3 + \int_0^3 \frac{2}{1+t^2}dt \right] = 2\left( -\frac{\ln 10}{3} + 2\arctan 3 \right)
$$


```python
from math import log, sqrt, atan
from typing import Callable
```

```python
def f(x: float) -> float:
    return log(1 + x) / x / sqrt(x)

def g(t: float) -> float:
    if t == 0:
        return 2.0
    return 2 * log(1 + t**2) / (t**2)
```

```python
def trapezoid_method(
    func: Callable[[float], float],
    a: float,
    b: float,
    epsilon: float = 1e-4,
    max_iterations: int = 20,
) -> float:
    """Метод трапеций"""
    prev_result = 0.0
    current_result = 0.0
    n = 1
    iteration = 0

    while iteration < max_iterations:
        prev_result = current_result
        h = (b - a) / n
        summation = 0.0
        for i in range(1, n):
            summation += func(a + i * h)
        current_result = h * ((func(a) + func(b)) / 2 + summation)

        if iteration > 0 and abs(current_result - prev_result) < epsilon:
            break

        n *= 2
        iteration += 1

    return current_result
```

```python
theory = - 2 * log(10) / 3 + 4 * atan(3)

# Границы интегрирования (0, 3)
a, b = 0, 3

# Погрешность 10^-4
epsilon = 1e-4

practice = trapezoid_method(g, a, b, epsilon, max_iterations=30)

delta = abs(theory - practice)

print(f"{theory=}")
print(f"{practice=}")
print(f"{delta=}")
```
