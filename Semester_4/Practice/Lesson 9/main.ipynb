# Лабораторная работа №5. Вариант № 18

Решить методами Гаусса и Якоби, найти $\lambda_{\min }, \lambda_{\max }$, определить число обусловленности матрицы $\mu=\|A\| \cdot\left\|A^{-1}\right\|$. Сделать вывод нормы вектора невязки для обоих методов.
$$
\left\{\begin{array}{l}
a_{11} x_1+a_{12} x_2+\ldots+a_{1 n} x_n=f_1 \\
\cdots \\
a_{n 1} x_1+a_{n 2} x_2+\ldots+a_{n n} x_n=f_n
\end{array}\right.
$$
$$
n=12, a_{i i}=1, a_{i j}=1 /\left(i^2+j\right)(i \neq j), f_i=1 / i

$$

Глобальные переменные

```python
N = 12
```

Генерирование матрицы

```python
def a_ij(i: float, j: float) -> float:
    """Функция элементов матрицы A для задания (вариант 18)"""
    i += 1
    j += 1
    if i == j:
        return 1
    return 1 / (i**2 + j)


def f_i(i: float, j: float) -> float:
    """Функция значений матрицы B для задания (вариант 18)"""
    i += 1
    j += 1
    return 1 / i


def generate_start_matrix(n: int = N) -> tuple[list[list[float]]]:
    """Создание матрицы A для задания (вариант 18)"""
    A = [[0.] * n for _ in range(n)]
    B = [0.] * n

    for i in range(n):
        for j in range(n):
            A[i][j] = a_ij(i, j)

    for i in range(n):
        B[i] = f_i(i, j)
        
    return A, B
```

Реализация метода Гаусса

```python
def method_gauss(values: list[list[float]], functions: list[float]) -> list[float]:
    n = len(functions)
    
    # Проверка на квадратность матрицы
    if len(values) != n:
        raise ValueError(f"Матрица должна быть квадратной! ({len(values)} != {n})")
    
    matrix = [row[:] + [functions[i]] for i, row in enumerate(values)]
    
    for i in range(n):
        max_row = i
        max_val = abs(matrix[i][i])
        for k in range(i + 1, n):
            if abs(matrix[k][i]) > max_val:
                max_val = abs(matrix[k][i])
                max_row = k
        
        # Если ведущий элемент близок к нулю - матрица вырождена
        if max_val < 1e-12:
            raise ValueError(f"Матрица вырождена! Ведущий элемент в столбце {i} равен нулю.")
        
        # Перестановка строк, если нужно
        if max_row != i:
            matrix[i], matrix[max_row] = matrix[max_row], matrix[i]
        
        # Делим на ведущий элемент
        current = matrix[i][i]
        for j in range(i, n + 1):
            matrix[i][j] /= current
        
        # Исключение переменной x_i из всех строк НИЖЕ
        for j in range(i + 1, n):
            factor = matrix[j][i]
            for k in range(i, n + 1):
                matrix[j][k] -= factor * matrix[i][k]
    
    x = [0.0] * n
    for i in range(n - 1, -1, -1):
        # Вычисляем сумму уже найденных переменных
        s = sum(matrix[i][j] * x[j] for j in range(i + 1, n))
        # Находим x_i
        x[i] = matrix[i][n] - s
    
    return x
```

```python
result = method_gauss(
    *generate_start_matrix(N)
)

for row in result:
    print(f"{row:.{5}f}")
```
