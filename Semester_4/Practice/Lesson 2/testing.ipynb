2) Выполните отделение корней графически и найдите один из корней методами простой итерации и Ньютона с абсолютной погрешностью 0,0001.

3) Найдите одно из решений системы уравнений методом Ньютона с абсолютной погрешностью 0,0001.
$$\begin{cases}
\sin (x-0.6) - y = 1.6\\
3x + \cos y = 0.9
\end{cases}$$

```python
import methods
from typing import Callable
from dataclasses import dataclass
from math import log, e, cos, tan
```

```python
@dataclass
class ExampleTask1:
    name: str
    func: Callable[[float], float]
    a: int
    b: int
    epsilon: float = 10e-5
    max_iteration: float = 10_000

    def __str__(self) -> str:
        return f"Ex. '{self.name}': a = {self.a}, b = {self.b}, \
epsilon = {self.epsilon}, max iteration = {self.max_iteration}"


@dataclass
class ExampleTask2:
    name: str
    func: Callable[[float], float]
    a: int
    b: int
    x0: int = None
    epsilon: float = 10e-5
    max_iteration: float = 10_000

    def __str__(self) -> str:
        return f"Ex. '{self.name}': a = {self.a}, b = {self.b}, x0 = {self.x0}, \
epsilon = {self.epsilon}, max iteration = {self.max_iteration}"
```

## Задание 1

```python
test_data = [
    ExampleTask1(
        name="Билет 1",
        func=lambda x: 3 * x**4 + 4 * x**3 - 12 * x**2 - 5,
        a=1,
        b=2,
        epsilon=0.0001,
    ),
    ExampleTask1(
        name="Билет 2",
        func=lambda x: 2 * x**4 - 9 * x**3 - 60 * x**2 + 1,
        a=0,
        b=1,
        epsilon=0.0001,
    ),
    ExampleTask1(
        name="Билет 3",
        func=lambda x: 3 * x**4 + 8 * x**3 + 6 * x**2 - 10,
        a=0,
        b=1,
        epsilon=0.0001,
    ),
    ExampleTask1(
        name="Билет 4", func=lambda x: x**5 + x**2 - 5, a=1, b=2, epsilon=0.0001
    ),
    ExampleTask1(
        name="Билет 5",
        func=lambda x: 3 * x**4 + 8 * x**3 + 6 * x**2 - 11,
        a=0.5,
        b=1,
        epsilon=0.0001,
    ),
    ExampleTask1(
        name="Билет 6", func=lambda x: x**4 - 18 * x**3 - 10, a=-1, b=0, epsilon=0.0001
    ),
    ExampleTask1(
        name="Билет 17", func=lambda x: x**4 - 18 * x - 10, a=-1, b=0, epsilon=0.0001
    ),
    ExampleTask1(
        name="Билет 18", func=lambda x: x**3 - 18 * x - 10, a=-1, b=0, epsilon=0.0001
    ),
]

for index, test in enumerate(test_data[-3:]):
    try:
        result_method_dichotomy = methods.method_dichotomy(
            func=test.func,
            a=test.a,
            b=test.b,
            epsilon=test.epsilon,
            max_iteration=test.max_iteration,
        )
    except Exception as ex:
        result_method_dichotomy = None
        print(ex)
    try:
        result_method_chord = methods.method_chord(
            func=test.func,
            a=test.a,
            b=test.b,
            epsilon=test.epsilon,
            max_iteration=test.max_iteration,
        )
    except Exception as ex:
        result_method_chord = None
        print(ex)

    print(f"""

{index + 1}) {test} 

Result Method Dichotomy: {result_method_dichotomy}
Result Method Chord: {result_method_chord}

""")
```
